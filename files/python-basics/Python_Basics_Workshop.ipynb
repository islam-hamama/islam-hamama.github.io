## 7. Functions

````python
def two_way_time(depth_m, velocity=2000):
    """Two-way travel time in seconds."""   # docstring
    return 2 * depth_m / velocity

t = two_way_time(1500)   # capture the returned value
print(t)

def acoustic_impedance(rho, v):
    return rho * v

print(acoustic_impedance(2.65, 3000))

# return != print :
#   print  -> only DISPLAYS text
#   return -> HANDS BACK a value you can store, reuse and test
# A function without return gives back None.
````

````python
def gardner_density(v, a=0.31, b=0.25):
    """Gardner's relation: seismic velocity -> bulk density."""
    return a * v ** b

print(gardner_density(3000))             # with defaults
print(gardner_density(3000, 0.23))       # override one
print(gardner_density(b=0.25, v=3000, a=0.31))  # keywords: any order

# CLASSIC BUG: never use a mutable default value
# def add_event(ev, catalog=[]):   <- DANGER: the list is shared between calls!
def add_event(ev, catalog=None):
    if catalog is None:
        catalog = []
    catalog.append(ev)
    return catalog

print(add_event("EQ-01"))   # ['EQ-01']
print(add_event("EQ-02"))   # ['EQ-02'] - correct!
````

## 12. Large Language Models from Python
Three ways to use an LLM: **chat apps** - **API from Python** (this section) - **agentic tools** like OpenCode (section 13).

````python
# Anthropic Claude
# %pip install anthropic
import os, getpass

if "ANTHROPIC_API_KEY" not in os.environ:
    os.environ["ANTHROPIC_API_KEY"] = getpass.getpass("Anthropic API key: ")

from anthropic import Anthropic
client = Anthropic()          # reads ANTHROPIC_API_KEY

msg = client.messages.create(
    model="claude-sonnet-4-5",   # check docs for current model names
    max_tokens=300,
    messages=[{"role": "user",
               "content": "Explain P-waves vs S-waves in 3 bullet points."}],
)
print(msg.content[0].text)
````

````python
# OpenAI - same idea, same shape
# %pip install openai
import os, getpass

if "OPENAI_API_KEY" not in os.environ:
    os.environ["OPENAI_API_KEY"] = getpass.getpass("OpenAI API key: ")

from openai import OpenAI
client = OpenAI()             # reads OPENAI_API_KEY

resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user",
               "content": "Explain P-waves vs S-waves in 3 bullet points."}],
)
print(resp.choices[0].message.content)
````

````python
# SOLUTION: Station class
class Station:
    def __init__(self, name, lat, lon):
        self.name = name
        self.lat = lat
        self.lon = lon

    def hemisphere(self):
        return "Northern" if self.lat >= 0 else "Southern"

    def __str__(self):
        return f"Station({self.name}: {self.lat}, {self.lon} - {self.hemisphere()})"


s1 = Station("HLG", 27.3, 33.8)
s2 = Station("CTA", -20.1, 146.3)
print(s1)
print(s2)
````

````python
# SOLUTION (skeleton): Earthquake Catalog Analyzer with save/load
import json

FILE = "catalog.json"

def load():
    try:
        with open(FILE) as f:
            return json.load(f)
    except FileNotFoundError:
        return []

def save(catalog):
    with open(FILE, "w") as f:
        json.dump(catalog, f, indent=2)

def add_event(catalog, eid, mag, depth, region):
    catalog.append({"id": eid, "mag": mag, "depth": depth, "region": region})

def strongest(catalog):
    return max(catalog, key=lambda e: e["mag"]) if catalog else None

def filter_by_mag(catalog, min_mag):
    return [e for e in catalog if e["mag"] >= min_mag]

catalog = load()
while True:
    print("\n1-add 2-list 3-strongest 4-filter 5-save 0-quit")
    choice = input("> ").strip()
    if choice == "1":
        add_event(catalog,
                  input("id: "), float(input("mag: ")),
                  float(input("depth: ")), input("region: "))
    elif choice == "2":
        for e in catalog:
            print(e["id"], f"M{e['mag']}", f"{e['depth']} km", e["region"])
    elif choice == "3":
        print(strongest(catalog))
    elif choice == "4":
        print(filter_by_mag(catalog, float(input("min mag: "))))
    elif choice == "5":
        save(catalog)
        print("saved.")
    elif choice == "0":
        save(catalog)
        break
````
